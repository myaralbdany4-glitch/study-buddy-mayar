```python
def add_task(tasks, name):
    if not isinstance(name, str):
        print("Invalid task name")
        return tasks

    if name.strip() == "" or len(name) > 30:
        print("Invalid task name")
        return tasks

    tasks.append(name)
    return tasks

tasks = []
add_task(tasks, "Review Python")
print(tasks)
```
